```python
# 3 and 4

"""
IPL Match Summary Script
------------------------
This script stores basic details of an IPL match innings
(team name, runs scored, overs played) in variables,
then prints them and calculates the run rate.
Variable names follow Python's snake_case convention.
"""

# Section 1: Create variables for the IPL match
team_name = "Mumbai Indians"   # str: name of the batting team
runs_scored = 187              # int: total runs scored in the innings
overs_played = 20.0          # float: overs faced by the team

# Section 2: Print the variables
print("Team Name:", team_name)
print("Runs Scored:", runs_scored)
print("Overs Played:", overs_played)

"""
Section 3: Calculate the run rate
Run rate = total runs scored / overs played.
This shows how many runs the team scored per over on average.
"""
run_rate = runs_scored / overs_played

# Print the run rate rounded to 2 decimal places
print("Run Rate:", round(run_rate, 2))
```
