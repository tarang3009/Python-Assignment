```python
# 2

def total_cart_amount(prices):
    total = 0.0
    for price in prices:
        # prices come as strings so convert each one first
        total += float(price)
    return total

# sample cart
cart = ['199.99', '49', '350.75']
print("Total cart amount:", total_cart_amount(cart))
```

```python
# 3

score = input("Enter your cricket score: ")
score = int(score)

if score >= 50 and score < 100:
    print("Half-century!")
elif score >= 100:
    print("Wonderful Century!")
else:
    print("Keep going!")
```

```python
# 4

is_premium = 'True'

# bool('False') would also give True because any non-empty string is True
# bool() only checks if the string is empty or not. Any non-empty string, including 'False', becomes True.
# so compare the string instead
is_premium = (is_premium == 'True')

print(is_premium)
print(type(is_premium))
```
