```python
# 1

def get_song_duration_per_minute(total_minutes, num_songs):
    try:
        result = total_minutes / num_songs
        print("Average song duration:", round(result, 2), "minutes")
    except ZeroDivisionError:
        print("Number of songs cannot be zero")
    finally:
        print("Calculation finished")

get_song_duration_per_minute(60, 15)
get_song_duration_per_minute(60, 0)
```

```python
# 2

total_amount = float(input("Enter total cart amount: "))
item_count = int(input("Enter number of items: "))

try:
    price_per_item = total_amount / item_count
    print("Price per item: ₹", round(price_per_item, 2))
except ZeroDivisionError:
    print("Item count cannot be zero. \nPlease add items to your cart.")
```

```python
# 3

class NoOffersApplied(Exception):
    pass

total_spend = float(input("Enter total spend: "))
offers = int(input("Enter number of offers applied: "))

try:
    if offers == 0:
        raise NoOffersApplied("No offers were applied, so cashback cannot be calculated.")
    average = total_spend / offers
    print("Average cashback per offer: ₹", round(average, 2))
except NoOffersApplied as e:
    print("Error:", e)
```

```python
# 4

def calculate_average_rating(total_rating, num_reviews):
    try:
        return total_rating / num_reviews
    except ZeroDivisionError:
        return "No reviews yet"
    except TypeError:
        return "Invalid input"
    finally:
        print("Thank you for using the calculator")

print(calculate_average_rating(500, 0))
```

```python
# 5

def safe_divide_for_zomato(bill_amount, num_people):
    try:
        share = bill_amount / num_people
    except ZeroDivisionError:
        print("Error: Number of people cannot be zero")
    else:
        print("Each person pays: ₹", round(share, 2))
    finally:
        print("Split calculation done")

safe_divide_for_zomato(900, 3)
safe_divide_for_zomato(900, 0)
```
