```python
# 3

import csv

rows = [
    ["match_id", "team1", "team2", "winner"],
    [1, "Mumbai Indians", "CSK", "CSK"],
    [2, "RCB", "KKR", "RCB"],
    [3, "Rajasthan Royals", "Gujarat Titans", "Gujarat Titans"],
    [4, "Delhi Capitals", "Mumbai Indians", "Mumbai Indians"]
]

with open("ipl_matches.csv", "w", newline="") as file:
    writer = csv.writer(file)
    writer.writerows(rows)

with open("ipl_matches.csv", "r") as file:
    reader = csv.DictReader(file)
    for row in reader:
        print("Match", row["match_id"], "winner:", row["winner"])
```

```python
# 4

import json

movies = [
    {"title": "Jawan", "year": 2023, "rating": 7.0},
    {"title": "Animal", "year": 2023, "rating": 6.2},
    {"title": "Pathaan", "year": 2023, "rating": 5.9}
]

with open("movies.json", "w") as file:
    json.dump(movies, file, indent=4)

with open("movies.json", "r") as file:
    data = json.load(file)

for movie in data:
    print(movie["title"], "-", movie["rating"])
```

```python
# 5

from pathlib import Path
import json

file_path = Path("my_fav_apps.json")

if not file_path.exists():
    apps = [
        {"name": "Instagram", "category": "Social Media"},
        {"name": "Zomato", "category": "Food Delivery"},
        {"name": "Paytm", "category": "Payments"}
    ]
    with open(file_path, "w") as file:
        json.dump(apps, file, indent=4)
    print("File created")
else:
    print("File already exists")
```
