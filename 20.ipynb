```python
# 1

class Playlist:
    def __init__(self):
        self._songs = []

    def add_song(self, song):
        self._songs.append(song)

    def show_playlist(self):
        print(self._songs)

my_playlist = Playlist()
my_playlist.add_song("Kesariya")
my_playlist.add_song("Apna Bana Le")
my_playlist.add_song("Tum Hi Ho")

my_playlist.show_playlist()
```

```python
# 2

class Product:
    def __init__(self, name, price):
        self.name = name
        self._price = price

    def get_price(self):
        return self._price

    def set_price(self, new_price):
        self._price = new_price

phone = Product("Redmi Note 13", 15999)

print("Price:", phone.get_price())

phone.set_price(14499)
print("Updated price:", phone.get_price())
```

```python
# 3

class Movie:
    def __init__(self, title, rating):
        self.title = title
        self.__rating = 0.0
        self.set_rating(rating)

    def get_rating(self):
        return self.__rating

    def set_rating(self, value):
        if 0 <= value <= 10:
            self.__rating = float(value)
        else:
            print("Error: Rating must be between 0 and 10")

movie = Movie("Jawan", 7.5)
print("Rating:", movie.get_rating())

movie.set_rating(12)
print("Rating:", movie.get_rating())

movie.set_rating(8.2)
print("Rating:", movie.get_rating())
```

```python
# 4

from abc import ABC, abstractmethod

class PaymentMethod(ABC):
    @abstractmethod
    def pay(self, amount):
        pass

class Paytm(PaymentMethod):
    def pay(self, amount):
        print(f"Paid ₹{amount} using Paytm wallet")

class PhonePe(PaymentMethod):
    def pay(self, amount):
        print(f"Paid ₹{amount} using PhonePe UPI")

paytm = Paytm()
phonepe = PhonePe()

paytm.pay(500)
phonepe.pay(750)
```
