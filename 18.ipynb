```python
# 1

import re

text = "Call me on +91-9876543210 or +91-9123456780. Wrong ones: +91-12345, 9876543210"
numbers = re.findall(r"\+91-\d{10}", text)
print(numbers)
```

```python
# 2

import re

def has_valid_date(text):
    if re.search(r"\b\d{2}/\d{2}/\d{4}\b", text):
        return True
    return False

print(has_valid_date("Order placed on 15/08/2024"))
print(has_valid_date("No date here"))
```

```python
# 3

import re

text = "Shirt costs Rs. 299, shoes cost Rs. 1500 and a cap is Rs. 250."
prices = re.findall(r"Rs\. (\d+)", text)
prices = [int(p) for p in prices]
print(prices)
print("Total:", sum(prices))
```

```python
# 4

import re

text = "Contact rahul@gmail.com or support@flipkart.com for help."
result = re.sub(r"[\w.+-]+@[\w-]+\.[\w.]+", "[hidden email]", text)
print(result)
```
