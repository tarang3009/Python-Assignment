```python
# 1

songs = ['Shape Of You', 'Blinding Lights', 'Levitating', 'Senorita']
cleaned = list(map(lambda title: title.lower(), songs))
print(cleaned)
```

```python
# 2

ratings = [4.2, 3.8, 4.5, 2.9, 3.5]
top_rated = list(filter(lambda r: r > 4.0, ratings))
print(top_rated)
```

```python
# 3

from functools import reduce

cart = [499, 1299, 299, 799]
total = reduce(lambda a, b: a + b, cart)
print("Total:", total)
```

```python
# 4

def format_followers(number):
    if number >= 1000000:
        return f"{number / 1000000:.1f}M"
    elif number >= 1000:
        return f"{number / 1000:.1f}K"
    else:
        return str(number)

followers = [950, 1500, 25000, 1200000]
print(list(map(format_followers, followers)))
```

```python
# 5

ipl_scores = [101, 98, 120, 77, 88]
even_scores = list(filter(lambda score: score % 2 == 0, ipl_scores))
print(even_scores)
```
