```python
# 1

def print_playlist_songs(songs):
    if len(songs) == 0:
        return
    print(songs[0])
    print_playlist_songs(songs[1:])

playlist = ["Kesariya", "Apna Bana Le", "Tum Hi Ho", "Raataan Lambiyan"]
print_playlist_songs(playlist)
```

```python
# 2

def count_unread_messages(group):
    total = group.get('count', 0)
    for sub in group.get('subgroups', []):
        total += count_unread_messages(sub)
    return total

chats = {
    'count': 5,
    'subgroups': [
        {'count': 3, 'subgroups': []},
        {'count': 2, 'subgroups': [
            {'count': 4, 'subgroups': []}
        ]}
    ]
}

print(count_unread_messages(chats))
```

```python
# 3

x = 'global'
def outer():
    x = 'outer'
    def inner():
        nonlocal x
        x = 'inner'
    inner()
    print('Inside outer:', x)
outer()
print('Outside:', x)
```

```python
# 4

def format_number_short(n):
    if n < 1000:
        return str(n)
    if n < 1000000:
        return str(round(n / 1000, 1)) + 'K'
    return str(round(n / 1000000, 1)) + 'M'

print(format_number_short(500))
print(format_number_short(1500))
print(format_number_short(1200000))
```

```python
# 5

def format_number_short(n, level=0):
    suffixes = ['', 'K', 'M']
    if n < 1000 or level == 2:
        if level == 0:
            return str(n)
        return str(round(n, 1)) + suffixes[level]
    return format_number_short(n / 1000, level + 1)

print(format_number_short(500))
print(format_number_short(1500))
print(format_number_short(1200000))
```
